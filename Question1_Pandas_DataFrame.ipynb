# Create, Save, and Explore a Student DataFrame Using Pandas

This project demonstrates how to create a student DataFrame using Pandas, save it as a CSV file, reload the data, and explore the DataFrame.

```python
import pandas as pd
```

## Step 2: Create the Student DataFrame

We create a dictionary containing student information and convert it into a Pandas DataFrame.

```python
data = {
    "StudentID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Ethan",
             "Fiona", "George", "Hannah", "Ivan", "Julia"],
    "Age": [20, 21, 19, 22, 20, 21, 23, 19, 22, 20],
    "Marks": [88, 75, 92, 65, 78, 85, 70, 95, 60, 80],
    "Grade": ["A", "B", "A", "C", "B", "A", "B", "A", "C", "B"]
}

df = pd.DataFrame(data)

df
```

## Step 3: Save the DataFrame as a CSV File

We save the DataFrame as `students.csv` so that the data can be stored and used later.

```python
df.to_csv("students.csv", index=False)

print("students.csv file saved successfully.")
```

## Step 4: Reload the CSV File

We read the saved `students.csv` file using Pandas and store it again in a DataFrame.

```python
df = pd.read_csv("students.csv")

df
```

## Step 5: Display the First 3 Rows

The `head(3)` method displays the first three rows of the DataFrame.

```python
df.head(3)
```

## Step 6: Statistical Summary

The `describe()` method provides statistical information about the numerical columns in the DataFrame, such as count, mean, standard deviation, minimum, and maximum.

```python
df.describe()
```

## Step 7: Check the Shape of the DataFrame

The `shape` attribute tells us the number of rows and columns in the DataFrame.

```python
df.shape
```

## Step 8: Check Data Types

The `dtypes` attribute shows the data type of each column in the DataFrame.

```python
df.dtypes
```

## Step 9: Display DataFrame Information

The `info()` method gives an overview of the DataFrame, including the number of rows, columns, non-null values, and data types.

```python
df.info()
```
